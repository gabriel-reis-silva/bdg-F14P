### Criação do Servidor de Matemática com FastMCP

Agora, vamos instanciar o `FastMCP` e definir as ferramentas para as quatro operações matemáticas básicas (soma, subtração, multiplicação e divisão).

```python
from fastmcp import FastMCP

# Cria uma instância do servidor FastMCP
mcp = FastMCP("MathServer")

@mcp.tool()
def somar(a: float, b: float) -> float:
    """Soma dois números."""
    return a + b

@mcp.tool()
def subtrair(a: float, b: float) -> float:
    """Subtrai o segundo número do primeiro."""
    return a - b

@mcp.tool()
def multiplicar(a: float, b: float) -> float:
    """Multiplica dois números."""
    return a * b

@mcp.tool()
def dividir(a: float, b: float) -> str:
    """Divide o primeiro número pelo segundo."""
    if b == 0:
        return "Erro: Divisão por zero não é permitida."
    return str(a / b)

# Para rodar o servidor em modo de desenvolvimento ou CLI localmente:
# mcp.run()
print("Servidor FastMCP 'MathServer' configurado com sucesso com as ferramentas: somar, subtrair, multiplicar e dividir!")
```

```python
# Importamos os módulos necessários do SDK do MCP
from mcp import ClientSession, StdioServerParameters
from mcp.client.stdio import stdio_client
import sys

# Vamos criar o código que atua 100% como cliente oficial chamando as ferramentas expostas via MCP
async def rodar_cliente_mcp():
    print("--- Cliente Conectando ao Servidor via FastMCP Core API ---")

    # No Python FastMCP, usamos o método `.call_tool` diretamente no objeto `mcp` para simular o protocolo do cliente
    resultado_mcp = await mcp.call_tool("multiplicar", arguments={"a": 3.0, "b": 7.0})

    # Exibe a resposta estruturada contendo o conteúdo de texto gerado pelo servidor MCP
    print("Resposta bruta recebida do servidor:")
    print(resultado_mcp)

    # Corrigindo o acesso aos dados do objeto ToolResult
    texto_resultado = resultado_mcp.content[0].text
    print(f"\nResultado Formatado: 3 x 7 = {texto_resultado}")

# Executa o cliente assíncrono
await rodar_cliente_mcp()
```
